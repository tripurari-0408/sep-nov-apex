```python
print("Age range:", df["Age"].min(), "to", df["Age"].max())
print("Quantity range:", df["Quantity"].min(), "to", df["Quantity"].max())
print("Unit Price range:", df["Unit_Price"].min(), "to", df["Unit_Price"].max())
print("Total Sales range:", df["Total_Sales"].min(), "to", df["Total_Sales"].max())
```

```python
df["Calculated_Sales"] = df["Quantity"] * df["Unit_Price"]
df["Sales_Difference"] = (df["Total_Sales"] - df["Calculated_Sales"]).abs()
mismatches = (df["Sales_Difference"] > 0.01).sum()

print("Rows with Total_Sales mismatch:", mismatches)
```

## 4. Data Cleaning & Transformation
The cleaning process below creates a separate analysis-ready dataframe.

```python
cleaned_df = df.drop(columns=["Calculated_Sales", "Sales_Difference"]).copy()
```

```python
# Standardize Order_Date
cleaned_df["Order_Date"] = pd.to_datetime(cleaned_df["Order_Date"], errors="coerce")
```

```python
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].median())
```

```python
cleaned_df["City"] = cleaned_df["City"].fillna("Unknown")
```

```python
cleaned_df = cleaned_df.drop_duplicates()
```

```python
# 5. Create an Age_Group feature
bins = [0, 18, 25, 35, 45, 60, 100]
labels = ["<18", "18-25", "26-35", "36-45", "46-60", "60+"]
```

```python
cleaned_df["Age_Group"] = pd.cut(
    cleaned_df["Age"],
    bins=bins,
    labels=labels,
    include_lowest=True)
```

```python
print("Final shape:", cleaned_df.shape)
print("Cleaning and transformation completed.")
```
